```python
from pathlib import Path
import json
import warnings
import time

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

SEED = 42
rng = np.random.default_rng(SEED)

# ================================================================
# 1. PROJECT PATHS
# ================================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based District-Wise Cattle Breeding Resource Analysis"
)

DATASET_PATH = (
    BASE_DIR / "Table_10.13_cowbulls_kept_AIcentres_0.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

print("=" * 75)
print("AIS-BASED CATTLE BREEDING RESOURCE ANALYSIS")
print("=" * 75)

# ================================================================
# 2. LOAD DATASET
# ================================================================

df = pd.read_csv(DATASET_PATH)

print("\nOriginal shape:", df.shape)

df = df.dropna(how="all")
df = df.dropna(axis=1, how="all")

DISTRICT_COL = "District/Year"

if DISTRICT_COL not in df.columns:
    raise ValueError(
        "Required District/Year column missing."
    )

df[DISTRICT_COL] = (
    df[DISTRICT_COL]
    .astype("string")
    .str.strip()
)

df = df.dropna(subset=[DISTRICT_COL])
df = df[df[DISTRICT_COL] != ""].copy()

# Exclude the statewide total from district-level training.
df = df[
    df[DISTRICT_COL].str.lower() != "punjab"
].copy()

year_cols = sorted(
    [c for c in df.columns if str(c).isdigit()],
    key=int
)

if not year_cols:
    raise ValueError("No historical year columns found.")

for col in year_cols:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# ================================================================
# 3. TRANSFORM DATASET
# ================================================================

long_df = df.melt(
    id_vars=[DISTRICT_COL],
    value_vars=year_cols,
    var_name="Year",
    value_name="Animal_Count"
)

long_df = long_df.rename(
    columns={DISTRICT_COL: "District"}
)

long_df["Year"] = long_df["Year"].astype(int)

long_df["Animal_Count"] = pd.to_numeric(
    long_df["Animal_Count"],
    errors="coerce"
)

# Preserve real zeros. Remove only missing observations.
long_df = long_df.dropna(
    subset=["Animal_Count"]
)

long_df = long_df[
    long_df["Animal_Count"] >= 0
].copy()

long_df = long_df.sort_values(
    ["District", "Year"]
).reset_index(drop=True)

print("\nDistricts:", long_df["District"].nunique())
print("Valid observations:", len(long_df))

# ================================================================
# 4. FEATURE ENGINEERING
# ================================================================

district_names = sorted(
    long_df["District"].unique()
)

district_mapping = {
    name: index
    for index, name in enumerate(district_names)
}

FEATURES = [
    "Year",
    "Lag_1",
    "Lag_2",
    "Lag_3",
    "Moving_Average_3",
    "Previous_Change",
    "District_Code"
]

records = []

for district, group in long_df.groupby("District"):

    values = dict(
        zip(group["Year"], group["Animal_Count"])
    )

    for year in sorted(values):

        # Require three consecutive prior years.
        required = [
            year - 1,
            year - 2,
            year - 3
        ]

        if not all(y in values for y in required):
            continue

        lag1 = float(values[year - 1])
        lag2 = float(values[year - 2])
        lag3 = float(values[year - 3])

        records.append({
            "District": district,
            "Year": int(year),
            "Lag_1": lag1,
            "Lag_2": lag2,
            "Lag_3": lag3,
            "Moving_Average_3": np.mean(
                [lag1, lag2, lag3]
            ),
            "Previous_Change": lag1 - lag2,
            "District_Code": district_mapping[district],
            "Animal_Count": float(values[year])
        })

model_df = pd.DataFrame(records)

if model_df.empty:
    raise ValueError(
        "No valid consecutive-year samples available."
    )

print("\nModel-ready observations:", len(model_df))

# ================================================================
# 5. CHRONOLOGICAL DATA SPLIT
# ================================================================

# Reproduce the earlier 65% / 80% year boundaries.
# Never tune hyperparameters on the test set.

years = sorted(model_df["Year"].unique())

if len(years) < 9:
    raise ValueError(
        "Insufficient distinct years for splitting."
    )

train_end_idx = max(
    1,
    int(len(years) * 0.65) - 1
)

val_end_idx = max(
    train_end_idx + 1,
    int(len(years) * 0.80) - 1
)

train_end = years[train_end_idx]
val_end = years[val_end_idx]

train_df = model_df[
    model_df["Year"] <= train_end
].copy()

val_df = model_df[
    (model_df["Year"] > train_end) &
    (model_df["Year"] <= val_end)
].copy()

test_df = model_df[
    model_df["Year"] > val_end
].copy()

if min(
    len(train_df),
    len(val_df),
    len(test_df)
) == 0:
    raise ValueError("A data split is empty.")

X_train = train_df[FEATURES]
y_train = train_df["Animal_Count"].to_numpy()

X_val = val_df[FEATURES]
y_val = val_df["Animal_Count"].to_numpy()

X_test = test_df[FEATURES]
y_test = test_df["Animal_Count"].to_numpy()

print("\nTraining samples:", len(train_df))
print("Validation samples:", len(val_df))
print("Testing samples:", len(test_df))
print("Train end:", train_end)
print("Validation end:", val_end)

# ================================================================
# 6. EVALUATION FUNCTIONS
# ================================================================

def evaluate(actual, predicted):

    actual = np.asarray(actual, dtype=float)
    predicted = np.asarray(predicted, dtype=float)

    result = {
        "MAE": float(
            mean_absolute_error(actual, predicted)
        ),
        "RMSE": float(
            np.sqrt(
                mean_squared_error(actual, predicted)
            )
        ),
        "R2": float(
            r2_score(actual, predicted)
        ) if (
            len(actual) > 1 and
            np.var(actual) > 0
        ) else None
    }

    return result


def build_model(params):

    return RandomForestRegressor(
        n_estimators=int(params["n_estimators"]),
        max_depth=int(params["max_depth"]),
        min_samples_split=int(
            params["min_samples_split"]
        ),
        min_samples_leaf=int(
            params["min_samples_leaf"]
        ),
        max_features=float(
            params["max_features"]
        ),
        random_state=SEED,
        n_jobs=-1
    )

# ================================================================
# 7. BASELINE RANDOM FOREST
# ================================================================

baseline_params = {
    "n_estimators": 250,
    "max_depth": 12,
    "min_samples_split": 2,
    "min_samples_leaf": 2,
    "max_features": 1.0
}

baseline_model = build_model(
    baseline_params
)

baseline_model.fit(
    X_train,
    y_train
)

baseline_val_pred = np.maximum(
    baseline_model.predict(X_val),
    0
)

baseline_val_mae = mean_absolute_error(
    y_val,
    baseline_val_pred
)

print("\nBaseline Validation MAE:", baseline_val_mae)

# ================================================================
# 8. AIS SEARCH SPACE
# ================================================================

# Each antibody encodes a candidate set of Random Forest
# hyperparameters.
#
# Affinity = 1 / (1 + validation MAE)
# Higher affinity means better validation performance.

PARAM_BOUNDS = {
    "n_estimators": (50, 300),
    "max_depth": (3, 20),
    "min_samples_split": (2, 15),
    "min_samples_leaf": (1, 10),
    "max_features": (0.4, 1.0)
}

PARAM_NAMES = list(PARAM_BOUNDS.keys())

INTEGER_PARAMS = {
    "n_estimators",
    "max_depth",
    "min_samples_split",
    "min_samples_leaf"
}

def decode_antibody(antibody):

    params = {}

    for index, name in enumerate(PARAM_NAMES):

        low, high = PARAM_BOUNDS[name]

        value = low + float(antibody[index]) * (
            high - low
        )

        if name in INTEGER_PARAMS:
            value = int(round(value))
        else:
            value = float(value)

        params[name] = value

    return params


def random_antibody():

    return rng.uniform(
        0,
        1,
        size=len(PARAM_NAMES)
    )


# ================================================================
# 9. AIS FITNESS FUNCTION
# ================================================================

fitness_cache = {}

def fitness(antibody):

    params = decode_antibody(antibody)

    key = tuple(
        params[name]
        for name in PARAM_NAMES
    )

    if key in fitness_cache:
        return fitness_cache[key]

    model = build_model(params)

    model.fit(
        X_train,
        y_train
    )

    predictions = np.maximum(
        model.predict(X_val),
        0
    )

    mae = float(
        mean_absolute_error(
            y_val,
            predictions
        )
    )

    affinity = 1.0 / (1.0 + mae)

    fitness_cache[key] = {
        "mae": mae,
        "affinity": affinity,
        "params": params
    }

    return fitness_cache[key]


# ================================================================
# 10. AIS CLONAL SELECTION ALGORITHM
# ================================================================

POPULATION_SIZE = 12
GENERATIONS = 15
ELITE_COUNT = 4
CLONES_PER_ELITE = 5
RANDOM_REPLACEMENTS = 3

population = [
    random_antibody()
    for _ in range(POPULATION_SIZE)
]

best_antibody = None
best_fitness = float("inf")

optimization_history = []

start_time = time.time()

print("\n" + "=" * 75)
print("STARTING AIS OPTIMIZATION")
print("=" * 75)

for generation in range(GENERATIONS):

    # Evaluate antibodies.
    scored = []

    for antibody in population:

        score = fitness(antibody)

        scored.append(
            (
                score["mae"],
                antibody.copy()
            )
        )

    scored.sort(
        key=lambda item: item[0]
    )

    generation_best = scored[0][0]

    if generation_best < best_fitness:

        best_fitness = generation_best

        best_antibody = (
            scored[0][1].copy()
        )

    # Select highest-affinity antibodies.
    elites = [
        antibody.copy()
        for _, antibody in scored[:ELITE_COUNT]
    ]

    clone_candidates = []

    # Clone and hypermutate selected antibodies.
    for rank, elite in enumerate(elites):

        # Better antibodies receive smaller mutations.
        mutation_sigma = (
            0.05 + 0.20 * (
                rank / max(1, ELITE_COUNT - 1)
            )
        )

        for _ in range(CLONES_PER_ELITE):

            clone = elite.copy()

            mutation = rng.normal(
                0,
                mutation_sigma,
                size=len(clone)
            )

            clone = np.clip(
                clone + mutation,
                0,
                1
            )

            clone_score = fitness(clone)["mae"]

            clone_candidates.append(
                (
                    clone_score,
                    clone.copy()
                )
            )

    # Combine parents and clones.
    candidates = scored + clone_candidates

    candidates.sort(
        key=lambda item: item[0]
    )

    # Retain the best antibodies.
    retained_count = (
        POPULATION_SIZE - RANDOM_REPLACEMENTS
    )

    population = [
        antibody.copy()
        for _, antibody in candidates[:retained_count]
    ]

    # Introduce new random antibodies for diversity.
    for _ in range(RANDOM_REPLACEMENTS):
        population.append(
            random_antibody()
        )

    # Preserve global best (elitism).
    population[0] = best_antibody.copy()

    optimization_history.append({
        "Generation": generation + 1,
        "Generation_Best_MAE": generation_best,
        "Global_Best_MAE": best_fitness,
        "Global_Best_Affinity": (
            1.0 / (1.0 + best_fitness)
        ),
        "Evaluated_Configurations": len(
            fitness_cache
        )
    })

    print(
        f"Generation {generation + 1:02d}/{GENERATIONS}"
        f" | Best MAE: {best_fitness:.5f}"
        f" | Evaluations: {len(fitness_cache)}"
    )

# Evaluate the final population, including random replacements.
for antibody in population:

    score = fitness(antibody)

    if score["mae"] < best_fitness:
        best_fitness = score["mae"]
        best_antibody = antibody.copy()

elapsed = time.time() - start_time

best_params = decode_antibody(
    best_antibody
)

print("\nAIS OPTIMIZATION COMPLETED")
print("Best parameters:", best_params)
print("Best validation MAE:", best_fitness)
print("Optimization time:", round(elapsed, 2), "seconds")

history_df = pd.DataFrame(
    optimization_history
)

history_df.to_csv(
    BASE_DIR / "ais_optimization_history.csv",
    index=False
)

# ================================================================
# 11. AIS CONVERGENCE GRAPH
# ================================================================

plt.figure(figsize=(10, 6))

plt.plot(
    history_df["Generation"],
    history_df["Global_Best_MAE"],
    marker="o",
    linewidth=2,
    label="AIS Best Validation MAE"
)

plt.axhline(
    baseline_val_mae,
    linestyle="--",
    label="Baseline Validation MAE"
)

plt.xlabel("Generation")
plt.ylabel("Validation MAE")
plt.title(
    "AIS Optimization Convergence",
    fontsize=14,
    fontweight="bold"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_convergence_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 12. TRAIN FINAL AIS MODEL
# ================================================================

# For fair test comparison, fit both models using only
# the original training split.
# Validation was used to select AIS parameters.

ais_model = build_model(
    best_params
)

ais_model.fit(
    X_train,
    y_train
)

ais_val_pred = np.maximum(
    ais_model.predict(X_val),
    0
)

ais_test_pred = np.maximum(
    ais_model.predict(X_test),
    0
)

baseline_test_pred = np.maximum(
    baseline_model.predict(X_test),
    0
)

# Save the optimized model.
joblib.dump(
    ais_model,
    BASE_DIR / "ais_model.pkl"
)

joblib.dump(
    {
        "features": FEATURES,
        "district_mapping": district_mapping,
        "train_end_year": train_end,
        "validation_end_year": val_end,
        "best_params": best_params,
        "model_type": "AIS-optimized Random Forest"
    },
    BASE_DIR / "ais_preprocessing.pkl"
)

print("\nAIS model saved successfully.")

# ================================================================
# 13. EVALUATE AIS AND BASELINE
# ================================================================

baseline_metrics = evaluate(
    y_test,
    baseline_test_pred
)

ais_metrics = evaluate(
    y_test,
    ais_test_pred
)

comparison_df = pd.DataFrame([
    {
        "Model": "Baseline Random Forest",
        **baseline_metrics
    },
    {
        "Model": "AIS Optimized Random Forest",
        **ais_metrics
    }
])

comparison_df.to_csv(
    BASE_DIR / "ais_comparison_results.csv",
    index=False
)

print("\nMODEL COMPARISON")
print(comparison_df.to_string(index=False))

# ================================================================
# 14. AIS ACCURACY / PERFORMANCE GRAPH
# ================================================================

# R2 is a regression metric, not classification accuracy.
# Negative R2 values must remain visible.

fig, axes = plt.subplots(
    1, 2,
    figsize=(14, 5)
)

sns.barplot(
    data=comparison_df,
    x="Model",
    y="R2",
    ax=axes[0]
)

axes[0].set_title("Test R² Score")
axes[0].set_ylabel("R²")
axes[0].tick_params(axis="x", rotation=12)
axes[0].axhline(0, color="black", linewidth=1)

sns.barplot(
    data=comparison_df,
    x="Model",
    y="MAE",
    ax=axes[1]
)

axes[1].set_title("Test Mean Absolute Error")
axes[1].set_ylabel("MAE (Lower is Better)")
axes[1].tick_params(axis="x", rotation=12)

plt.suptitle(
    "AIS Model Prediction Performance",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 15. AIS CORRELATION HEATMAP
# ================================================================

heatmap_columns = [
    "Year",
    "Lag_1",
    "Lag_2",
    "Lag_3",
    "Moving_Average_3",
    "Previous_Change",
    "Animal_Count"
]

correlation = model_df[
    heatmap_columns
].corr()

plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    square=True
)

plt.title(
    "AIS Project - Feature Correlation Heatmap",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 16. AIS COMPARISON GRAPH
# ================================================================

fig, axes = plt.subplots(
    1, 3,
    figsize=(17, 5)
)

metric_info = [
    ("MAE", "Mean Absolute Error"),
    ("RMSE", "Root Mean Squared Error"),
    ("R2", "R² Score")
]

for ax, (metric, title) in zip(
    axes,
    metric_info
):

    sns.barplot(
        data=comparison_df,
        x="Model",
        y=metric,
        ax=ax
    )

    ax.set_title(title)
    ax.tick_params(axis="x", rotation=15)

    if metric == "R2":
        ax.axhline(
            0,
            color="black",
            linewidth=1
        )

plt.suptitle(
    "Baseline Random Forest vs AIS Optimization",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 17. AIS RESULT CSV
# ================================================================

result_df = test_df[
    ["District", "Year", "Animal_Count"]
].copy()

result_df = result_df.rename(
    columns={
        "Animal_Count": "Actual_Count"
    }
)

result_df["Baseline_Prediction"] = (
    baseline_test_pred
)

result_df["AIS_Prediction"] = (
    ais_test_pred
)

result_df["AIS_Absolute_Error"] = np.abs(
    result_df["Actual_Count"] -
    result_df["AIS_Prediction"]
)

result_df["AIS_Squared_Error"] = (
    result_df["Actual_Count"] -
    result_df["AIS_Prediction"]
) ** 2

result_df = result_df.sort_values(
    ["Year", "District"]
).reset_index(drop=True)

result_df.to_csv(
    BASE_DIR / "ais_result.csv",
    index=False
)

print("\nAIS RESULTS")
print(result_df.head(15).to_string(index=False))

# ================================================================
# 18. AIS RESULT GRAPH
# ================================================================

fig, axes = plt.subplots(
    1, 2,
    figsize=(16, 6)
)

# Actual vs predicted scatter.
sns.scatterplot(
    data=result_df,
    x="Actual_Count",
    y="AIS_Prediction",
    ax=axes[0],
    alpha=0.7
)

max_value = max(
    result_df["Actual_Count"].max(),
    result_df["AIS_Prediction"].max(),
    1
)

axes[0].plot(
    [0, max_value],
    [0, max_value],
    "r--",
    label="Perfect Prediction"
)

axes[0].set_title(
    "AIS Actual vs Predicted"
)
axes[0].legend()

# Year-wise test results.
year_summary = result_df.groupby(
    "Year",
    as_index=False
)[
    [
        "Actual_Count",
        "Baseline_Prediction",
        "AIS_Prediction"
    ]
].sum()

axes[1].plot(
    year_summary["Year"],
    year_summary["Actual_Count"],
    marker="o",
    label="Actual"
)

axes[1].plot(
    year_summary["Year"],
    year_summary["Baseline_Prediction"],
    marker="s",
    label="Baseline"
)

axes[1].plot(
    year_summary["Year"],
    year_summary["AIS_Prediction"],
    marker="^",
    label="AIS"
)

axes[1].set_title(
    "Year-wise Actual vs Predicted"
)
axes[1].set_xlabel("Year")
axes[1].set_ylabel("Animal Count")
axes[1].legend()

plt.suptitle(
    "AIS Prediction Results",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 19. DISTRICT-WISE AIS PERFORMANCE
# ================================================================

district_results = []

for district, group in result_df.groupby("District"):

    actual = group["Actual_Count"].to_numpy()
    predicted = group["AIS_Prediction"].to_numpy()

    district_results.append({
        "District": district,
        "Observations": len(group),
        **evaluate(actual, predicted)
    })

district_performance = pd.DataFrame(
    district_results
)

district_performance.to_csv(
    BASE_DIR / "ais_district_performance.csv",
    index=False
)

print("\nDISTRICT-WISE PERFORMANCE")
print(district_performance.to_string(index=False))

# ================================================================
# 20. FUTURE AIS PREDICTION
# ================================================================

# Forecast 2019-2023 using recursive prediction.
#
# Only districts with observed 2016, 2017 and 2018
# counts are eligible.
#
# These are historical out-of-sample projections,
# not verified observations.

LAST_YEAR = max(int(y) for y in year_cols)
FORECAST_HORIZON = 5

prediction_records = []
skipped_districts = []

for district, group in long_df.groupby("District"):

    history = {
        int(row.Year): float(row.Animal_Count)
        for row in group.itertuples()
    }

    required = [
        LAST_YEAR - 2,
        LAST_YEAR - 1,
        LAST_YEAR
    ]

    if not all(y in history for y in required):
        skipped_districts.append(district)
        continue

    for year in range(
        LAST_YEAR + 1,
        LAST_YEAR + FORECAST_HORIZON + 1
    ):

        lag1 = history[year - 1]
        lag2 = history[year - 2]
        lag3 = history[year - 3]

        feature_values = {
            "Year": year,
            "Lag_1": lag1,
            "Lag_2": lag2,
            "Lag_3": lag3,
            "Moving_Average_3": np.mean(
                [lag1, lag2, lag3]
            ),
            "Previous_Change": lag1 - lag2,
            "District_Code": district_mapping[district]
        }

        feature_df = pd.DataFrame(
            [feature_values]
        )[FEATURES]

        prediction = float(
            ais_model.predict(feature_df)[0]
        )

        prediction = max(0.0, prediction)

        # Use the prediction as a future lag.
        history[year] = prediction

        prediction_records.append({
            "District": district,
            "Year": year,
            "AIS_Predicted_Count": prediction,
            "AIS_Rounded_Count": int(
                round(prediction)
            ),
            "Model": "AIS Optimized Random Forest",
            "Prediction_Type": "Recursive forecast"
        })

prediction_df = pd.DataFrame(
    prediction_records,
    columns=[
        "District",
        "Year",
        "AIS_Predicted_Count",
        "AIS_Rounded_Count",
        "Model",
        "Prediction_Type"
    ]
)

prediction_df.to_csv(
    BASE_DIR / "ais_prediction.csv",
    index=False
)

print("\nAIS FUTURE PREDICTIONS")
print(prediction_df.head(15).to_string(index=False))

if skipped_districts:
    print("\nDistricts excluded due to missing final-year data:")
    print(skipped_districts)

# ================================================================
# 21. AIS PREDICTION GRAPH
# ================================================================

if not prediction_df.empty:

    fig, axes = plt.subplots(
        1, 2,
        figsize=(16, 6)
    )

    future_summary = prediction_df.groupby(
        "Year",
        as_index=False
    )["AIS_Predicted_Count"].sum()

    sns.lineplot(
        data=future_summary,
        x="Year",
        y="AIS_Predicted_Count",
        marker="o",
        ax=axes[0]
    )

    axes[0].set_title(
        "AIS Forecast by Year (Included Districts)"
    )
    axes[0].set_ylabel("Predicted Animal Count")

    final_year = prediction_df["Year"].max()

    district_forecast = prediction_df[
        prediction_df["Year"] == final_year
    ].sort_values(
        "AIS_Predicted_Count",
        ascending=False
    )

    sns.barplot(
        data=district_forecast,
        x="AIS_Predicted_Count",
        y="District",
        ax=axes[1]
    )

    axes[1].set_title(
        f"AIS District Forecast - {final_year}"
    )
    axes[1].set_xlabel("Predicted Animal Count")

    plt.suptitle(
        "AIS-Based Cattle Breeding Resource Forecast",
        fontsize=15,
        fontweight="bold"
    )

    plt.tight_layout()

    plt.savefig(
        BASE_DIR / "ais_prediction_graph.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

else:
    print(
        "No eligible districts for prediction graph."
    )

# ================================================================
# 22. SAVE AIS CONFIGURATION YAML
# ================================================================

config = {
    "project": {
        "name": "AI-Based District-Wise Cattle Breeding Resource Analysis",
        "algorithm": "Artificial Immune System",
        "ais_variant": "Clonal Selection Algorithm",
        "dataset": DATASET_PATH.name,
        "output_directory": str(BASE_DIR)
    },
    "dataset": {
        "first_year": int(min(years)),
        "last_year": int(max(years)),
        "target": "Animal_Count",
        "features": FEATURES,
        "district_mapping": district_mapping
    },
    "splitting": {
        "type": "chronological",
        "train_end_year": int(train_end),
        "validation_end_year": int(val_end),
        "training_records": int(len(train_df)),
        "validation_records": int(len(val_df)),
        "testing_records": int(len(test_df))
    },
    "ais": {
        "population_size": POPULATION_SIZE,
        "generations": GENERATIONS,
        "elite_count": ELITE_COUNT,
        "clones_per_elite": CLONES_PER_ELITE,
        "random_replacements": RANDOM_REPLACEMENTS,
        "fitness": "Validation MAE",
        "best_validation_mae": float(best_fitness),
        "best_parameters": best_params,
        "evaluated_configurations": len(fitness_cache)
    },
    "forecast": {
        "start_year": int(LAST_YEAR + 1),
        "end_year": int(LAST_YEAR + FORECAST_HORIZON),
        "excluded_districts": skipped_districts
    }
}

with open(
    BASE_DIR / "ais_config.yaml",
    "w",
    encoding="utf-8"
) as f:

    yaml.safe_dump(
        config,
        f,
        sort_keys=False,
        allow_unicode=True
    )

# ================================================================
# 23. SAVE AIS METRICS JSON
# ================================================================

metrics = {
    "project": "AIS-Based Cattle Breeding Resource Analysis",
    "algorithm": "Clonal Selection",
    "optimization_objective": "Minimize validation MAE",
    "baseline_validation_mae": float(
        baseline_val_mae
    ),
    "ais_validation_mae": float(
        mean_absolute_error(y_val, ais_val_pred)
    ),
    "baseline_test_metrics": baseline_metrics,
    "ais_test_metrics": ais_metrics,
    "best_parameters": best_params,
    "optimization_generations": GENERATIONS,
    "unique_configurations_evaluated": len(
        fitness_cache
    ),
    "optimization_seconds": float(elapsed),
    "test_records": int(len(test_df)),
    "forecast_records": int(len(prediction_df)),
    "notes": [
        "AIS optimized Random Forest hyperparameters.",
        "Test data was not used in AIS fitness evaluation.",
        "R2 is a regression score, not classification accuracy.",
        "Missing observations were not converted to zero.",
        "Recursive forecasts are experimental projections."
    ]
}

with open(
    BASE_DIR / "ais_metrics.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metrics,
        f,
        indent=4
    )

# ================================================================
# 24. OUTPUT FILE VERIFICATION
# ================================================================

output_files = [
    "ais_model.pkl",
    "ais_preprocessing.pkl",
    "ais_config.yaml",
    "ais_metrics.json",
    "ais_accuracy_graph.png",
    "ais_heatmap.png",
    "ais_comparison_graph.png",
    "ais_comparison_results.csv",
    "ais_result.csv",
    "ais_result_graph.png",
    "ais_prediction.csv",
    "ais_prediction_graph.png",
    "ais_convergence_graph.png",
    "ais_optimization_history.csv",
    "ais_district_performance.csv"
]

print("\n" + "=" * 75)
print("AIS OUTPUT FILE VERIFICATION")
print("=" * 75)

for filename in output_files:

    path = BASE_DIR / filename

    if path.exists():
        print(f"[SAVED] {filename}")
    else:
        print(f"[MISSING] {filename}")

print("\n" + "=" * 75)
print("FINAL AIS RESULTS")
print("=" * 75)

print("\nBest Hyperparameters:")
print(json.dumps(best_params, indent=4))

print("\nBaseline Test Metrics:")
print(json.dumps(baseline_metrics, indent=4))

print("\nAIS Test Metrics:")
print(json.dumps(ais_metrics, indent=4))

print("\nOutput Directory:")
print(BASE_DIR)

print("\nAIS PROJECT EXECUTION COMPLETED SUCCESSFULLY")
```
