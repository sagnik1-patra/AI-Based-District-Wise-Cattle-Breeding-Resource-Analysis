```python
from pathlib import Path
import json
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import tensorflow as tf

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

# ================================================================
# 1. PROJECT PATHS
# ================================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based District-Wise Cattle Breeding Resource Analysis"
)

DATASET_PATH = (
    BASE_DIR / "Table_10.13_cowbulls_kept_AIcentres_0.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

print("=" * 70)
print("CATTLE BREEDING RESOURCE ANALYSIS")
print("RESULTS AND VISUALIZATIONS")
print("=" * 70)

# ================================================================
# 2. VERIFY REQUIRED FILES
# ================================================================

required_files = [
    DATASET_PATH,
    BASE_DIR / "cattle_breeding_model.pkl",
    BASE_DIR / "cattle_breeding_model.h5",
    BASE_DIR / "preprocessing.pkl",
    BASE_DIR / "metrics.json"
]

for file in required_files:
    if not file.exists():
        raise FileNotFoundError(
            f"Required file missing:\n{file}\n"
            "Run the previous model-training code first."
        )

# ================================================================
# 3. LOAD TRAINED MODELS
# ================================================================

rf_model = joblib.load(
    BASE_DIR / "cattle_breeding_model.pkl"
)

nn_model = tf.keras.models.load_model(
    str(BASE_DIR / "cattle_breeding_model.h5"),
    compile=False
)

preprocessing = joblib.load(
    BASE_DIR / "preprocessing.pkl"
)

with open(BASE_DIR / "metrics.json", "r") as f:
    saved_metrics = json.load(f)

scaler = preprocessing["scaler"]
district_mapping = preprocessing["district_mapping"]
FEATURES = preprocessing["features"]

selected_model = saved_metrics["selected_model"]

print("\nModels loaded successfully.")
print("Selected Model:", selected_model)

# ================================================================
# 4. LOAD AND CLEAN ORIGINAL DATA
# ================================================================

df = pd.read_csv(DATASET_PATH)

df = df.dropna(how="all")
df = df.dropna(axis=1, how="all")

district_col = "District/Year"

if district_col not in df.columns:
    raise ValueError("District/Year column missing.")

df[district_col] = (
    df[district_col]
    .astype("string")
    .str.strip()
)

df = df.dropna(subset=[district_col])
df = df[df[district_col] != ""]

# Exclude Punjab total from district-level modeling.
df = df[
    df[district_col].str.lower() != "punjab"
].copy()

year_cols = sorted(
    [c for c in df.columns if str(c).isdigit()],
    key=int
)

for col in year_cols:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

long_df = df.melt(
    id_vars=[district_col],
    value_vars=year_cols,
    var_name="Year",
    value_name="Animal_Count"
)

long_df = long_df.rename(
    columns={district_col: "District"}
)

long_df["Year"] = long_df["Year"].astype(int)

long_df["Animal_Count"] = pd.to_numeric(
    long_df["Animal_Count"],
    errors="coerce"
)

# Missing values are not assumed to be zero.
long_df = long_df.dropna(
    subset=["Animal_Count"]
)

long_df = long_df.sort_values(
    ["District", "Year"]
).reset_index(drop=True)

print("\nHistorical observations:", len(long_df))
print("Districts:", long_df["District"].nunique())

# ================================================================
# 5. RECREATE LAG FEATURES
# ================================================================

records = []

for district, group in long_df.groupby("District"):

    if district not in district_mapping:
        continue

    values = dict(
        zip(group["Year"], group["Animal_Count"])
    )

    for year in sorted(values):

        previous_years = [
            year - 1,
            year - 2,
            year - 3
        ]

        if not all(
            y in values for y in previous_years
        ):
            continue

        lag1 = float(values[year - 1])
        lag2 = float(values[year - 2])
        lag3 = float(values[year - 3])

        records.append({
            "District": district,
            "Year": year,
            "Lag_1": lag1,
            "Lag_2": lag2,
            "Lag_3": lag3,
            "Moving_Average_3": np.mean(
                [lag1, lag2, lag3]
            ),
            "Previous_Change": lag1 - lag2,
            "District_Code": district_mapping[district],
            "Animal_Count": float(values[year])
        })

model_df = pd.DataFrame(records)

if model_df.empty:
    raise ValueError(
        "No valid consecutive-year records available."
    )

# ================================================================
# 6. RECREATE ORIGINAL CHRONOLOGICAL TEST SET
# ================================================================

validation_end_year = int(
    preprocessing["validation_end_year"]
)

test_df = model_df[
    model_df["Year"] > validation_end_year
].copy()

if len(test_df) < 2:
    raise ValueError(
        "Insufficient test observations."
    )

X_test = test_df[FEATURES]
y_test = test_df["Animal_Count"].to_numpy()

X_test_scaled = scaler.transform(X_test)

# ================================================================
# 7. GENERATE MODEL PREDICTIONS
# ================================================================

rf_pred = np.maximum(
    rf_model.predict(X_test),
    0
)

nn_pred = np.maximum(
    nn_model.predict(
        X_test_scaled,
        verbose=0
    ).flatten(),
    0
)

if selected_model == "Random_Forest":
    best_pred = rf_pred
elif selected_model == "Neural_Network":
    best_pred = nn_pred
else:
    raise ValueError(
        f"Unknown selected model: {selected_model}"
    )

# ================================================================
# 8. EVALUATION FUNCTIONS
# ================================================================

def evaluate(actual, predicted):

    actual = np.asarray(actual, dtype=float)
    predicted = np.asarray(predicted, dtype=float)

    mae = mean_absolute_error(actual, predicted)

    rmse = np.sqrt(
        mean_squared_error(actual, predicted)
    )

    r2 = r2_score(actual, predicted)

    return {
        "MAE": float(mae),
        "RMSE": float(rmse),
        "R2": float(r2)
    }


rf_metrics = evaluate(y_test, rf_pred)
nn_metrics = evaluate(y_test, nn_pred)

comparison_df = pd.DataFrame([
    {
        "Model": "Random Forest",
        **rf_metrics
    },
    {
        "Model": "Neural Network",
        **nn_metrics
    }
])

comparison_df.to_csv(
    BASE_DIR / "comparison_results.csv",
    index=False
)

print("\nMODEL PERFORMANCE")
print(comparison_df.to_string(index=False))

# ================================================================
# 9. ACCURACY / PERFORMANCE GRAPH
# ================================================================

# R2 is a regression goodness-of-fit measure.
# It is not classification accuracy.
# R2 can be negative and is not capped below by zero.

fig, axes = plt.subplots(
    1, 2,
    figsize=(13, 5)
)

sns.barplot(
    data=comparison_df,
    x="Model",
    y="R2",
    ax=axes[0]
)

axes[0].axhline(
    0,
    color="black",
    linewidth=1
)

axes[0].set_title(
    "Model Performance (Test R²)"
)
axes[0].set_ylabel("R² Score")
axes[0].set_xlabel("Model")

for i, value in enumerate(comparison_df["R2"]):
    axes[0].annotate(
        f"{value:.3f}",
        (i, value),
        xytext=(0, 5 if value >= 0 else -15),
        textcoords="offset points",
        ha="center"
    )

sns.barplot(
    data=comparison_df,
    x="Model",
    y="MAE",
    ax=axes[1]
)

axes[1].set_title(
    "Mean Absolute Error (Lower is Better)"
)
axes[1].set_ylabel("MAE")
axes[1].set_xlabel("Model")

plt.suptitle(
    "Cattle Breeding Resource Prediction Performance",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: accuracy_graph.png")

# ================================================================
# 10. CORRELATION HEATMAP
# ================================================================

# Use the numeric training features and target.
# Exclude District_Code because numeric codes are arbitrary.
# Correlations are descriptive, not causal.

heatmap_columns = [
    "Year",
    "Lag_1",
    "Lag_2",
    "Lag_3",
    "Moving_Average_3",
    "Previous_Change",
    "Animal_Count"
]

corr = model_df[
    heatmap_columns
].corr()

plt.figure(figsize=(11, 8))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    square=True
)

plt.title(
    "Correlation Heatmap of Cattle Breeding Features",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: heatmap.png")

# ================================================================
# 11. MODEL COMPARISON GRAPH
# ================================================================

fig, axes = plt.subplots(
    1, 3,
    figsize=(16, 5)
)

metrics_to_plot = [
    ("MAE", "Mean Absolute Error"),
    ("RMSE", "Root Mean Squared Error"),
    ("R2", "R² Score")
]

for ax, (metric, title) in zip(
    axes,
    metrics_to_plot
):

    sns.barplot(
        data=comparison_df,
        x="Model",
        y=metric,
        ax=ax
    )

    ax.set_title(title)
    ax.set_xlabel("Model")
    ax.set_ylabel(metric)
    ax.tick_params(
        axis="x",
        rotation=10
    )

    if metric == "R2":
        ax.axhline(
            0,
            color="black",
            linewidth=1
        )

plt.suptitle(
    "Random Forest vs Neural Network",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: comparison_graph.png")
print("Saved: comparison_results.csv")

# ================================================================
# 12. GENERATE RESULT.CSV
# ================================================================

result_df = test_df[
    ["District", "Year", "Animal_Count"]
].copy()

result_df = result_df.rename(
    columns={
        "Animal_Count": "Actual_Count"
    }
)

result_df["RF_Prediction"] = rf_pred
result_df["NN_Prediction"] = nn_pred
result_df["Selected_Prediction"] = best_pred

result_df["Absolute_Error"] = np.abs(
    result_df["Actual_Count"] -
    result_df["Selected_Prediction"]
)

result_df["Squared_Error"] = (
    result_df["Actual_Count"] -
    result_df["Selected_Prediction"]
) ** 2

result_df["Selected_Model"] = selected_model

result_df = result_df.sort_values(
    ["Year", "District"]
).reset_index(drop=True)

result_df.to_csv(
    BASE_DIR / "result.csv",
    index=False
)

print("\nRESULT CSV")
print(result_df.head(15).to_string(index=False))

print("\nSaved: result.csv")

# ================================================================
# 13. RESULT GRAPH
# ================================================================

fig, axes = plt.subplots(
    1, 2,
    figsize=(15, 6)
)

# Actual vs predicted scatter
sns.scatterplot(
    data=result_df,
    x="Actual_Count",
    y="Selected_Prediction",
    ax=axes[0],
    alpha=0.7
)

max_count = max(
    result_df["Actual_Count"].max(),
    result_df["Selected_Prediction"].max(),
    1
)

axes[0].plot(
    [0, max_count],
    [0, max_count],
    "r--",
    label="Perfect Prediction"
)

axes[0].set_title(
    "Actual vs Predicted Animal Count"
)
axes[0].legend()

# Year-wise total across available test observations
year_summary = result_df.groupby(
    "Year",
    as_index=False
)[
    ["Actual_Count", "Selected_Prediction"]
].sum()

axes[1].plot(
    year_summary["Year"],
    year_summary["Actual_Count"],
    marker="o",
    label="Actual"
)

axes[1].plot(
    year_summary["Year"],
    year_summary["Selected_Prediction"],
    marker="s",
    label="Predicted"
)

axes[1].set_title(
    "Year-wise Test Results"
)

axes[1].set_xlabel("Year")
axes[1].set_ylabel("Animal Count")
axes[1].legend()

plt.suptitle(
    "Cattle Breeding Resource Prediction Results",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: result_graph.png")

# ================================================================
# 14. GENERATE FUTURE PREDICTION.CSV
# ================================================================

# Recursive forecasts from 2019 to 2023.
# Only districts with valid observations in 2016-2018
# can be initialized without inventing missing data.
#
# These are retrospective model forecasts, not actual counts.

last_year = max(int(y) for y in year_cols)
forecast_horizon = 5

prediction_records = []
skipped_districts = []

def predict_one(feature_values):

    X = pd.DataFrame(
        [feature_values]
    )[FEATURES]

    if selected_model == "Random_Forest":
        pred = rf_model.predict(X)[0]
    else:
        X_scaled = scaler.transform(X)

        pred = nn_model.predict(
            X_scaled,
            verbose=0
        )[0][0]

    return max(0.0, float(pred))


for district, group in long_df.groupby("District"):

    if district not in district_mapping:
        continue

    history = {
        int(row.Year): float(row.Animal_Count)
        for row in group.itertuples()
    }

    needed = [
        last_year - 2,
        last_year - 1,
        last_year
    ]

    if not all(y in history for y in needed):
        skipped_districts.append(district)
        continue

    for year in range(
        last_year + 1,
        last_year + forecast_horizon + 1
    ):

        lag1 = history[year - 1]
        lag2 = history[year - 2]
        lag3 = history[year - 3]

        features = {
            "Year": year,
            "Lag_1": lag1,
            "Lag_2": lag2,
            "Lag_3": lag3,
            "Moving_Average_3": np.mean(
                [lag1, lag2, lag3]
            ),
            "Previous_Change": lag1 - lag2,
            "District_Code": district_mapping[district]
        }

        prediction = predict_one(features)

        history[year] = prediction

        prediction_records.append({
            "District": district,
            "Year": year,
            "Predicted_Animal_Count": prediction,
            "Rounded_Prediction": int(round(prediction)),
            "Model": selected_model,
            "Prediction_Type": "Recursive forecast"
        })

prediction_df = pd.DataFrame(
    prediction_records,
    columns=[
        "District",
        "Year",
        "Predicted_Animal_Count",
        "Rounded_Prediction",
        "Model",
        "Prediction_Type"
    ]
)

prediction_df.to_csv(
    BASE_DIR / "prediction.csv",
    index=False
)

print("\nPREDICTION CSV")
print(prediction_df.head(15).to_string(index=False))

print("\nSaved: prediction.csv")

if skipped_districts:
    print(
        "\nExcluded from forecasting due to missing "
        "2016-2018 observations:"
    )
    print(skipped_districts)

# ================================================================
# 15. PREDICTION GRAPH
# ================================================================

if not prediction_df.empty:

    fig, axes = plt.subplots(
        1, 2,
        figsize=(16, 6)
    )

    # Year-wise aggregate prediction
    future_summary = prediction_df.groupby(
        "Year",
        as_index=False
    )["Predicted_Animal_Count"].sum()

    sns.lineplot(
        data=future_summary,
        x="Year",
        y="Predicted_Animal_Count",
        marker="o",
        ax=axes[0]
    )

    axes[0].set_title(
        "Forecasted Animal Counts (Included Districts)"
    )
    axes[0].set_ylabel("Predicted Animal Count")

    # District-wise forecast at the final forecast year
    final_year = prediction_df["Year"].max()

    district_forecast = prediction_df[
        prediction_df["Year"] == final_year
    ].sort_values(
        "Predicted_Animal_Count",
        ascending=False
    )

    sns.barplot(
        data=district_forecast,
        x="Predicted_Animal_Count",
        y="District",
        ax=axes[1]
    )

    axes[1].set_title(
        f"District-wise Forecast for {final_year}"
    )
    axes[1].set_xlabel("Predicted Animal Count")

    plt.suptitle(
        "District-Wise Cattle Breeding Resource Forecast",
        fontsize=15,
        fontweight="bold"
    )

    plt.tight_layout()

    plt.savefig(
        BASE_DIR / "prediction_graph.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

else:
    print(
        "No eligible districts for forecasting. "
        "Prediction graph not generated."
    )

# ================================================================
# 16. DISTRICT-WISE PERFORMANCE REPORT
# ================================================================

district_records = []

for district, group in result_df.groupby("District"):

    actual = group["Actual_Count"].to_numpy()
    predicted = group["Selected_Prediction"].to_numpy()

    district_records.append({
        "District": district,
        "Observations": len(group),
        "MAE": mean_absolute_error(
            actual,
            predicted
        ),
        "RMSE": np.sqrt(
            mean_squared_error(actual, predicted)
        ),
        "R2": (
            r2_score(actual, predicted)
            if len(actual) >= 2
            and np.var(actual) > 0
            else np.nan
        )
    })

district_metrics_df = pd.DataFrame(
    district_records
)

district_metrics_df.to_csv(
    BASE_DIR / "district_performance.csv",
    index=False
)

print("\nDISTRICT-WISE PERFORMANCE")
print(district_metrics_df.to_string(index=False))

# ================================================================
# 17. SAVE UPDATED METRICS.JSON
# ================================================================

visualization_metrics = {
    "project": "AI-Based District-Wise Cattle Breeding Resource Analysis",
    "selected_model": selected_model,
    "test_records": int(len(result_df)),
    "model_performance": {
        "Random_Forest": rf_metrics,
        "Neural_Network": nn_metrics
    },
    "forecast": {
        "start_year": int(last_year + 1),
        "end_year": int(last_year + forecast_horizon),
        "forecast_records": int(len(prediction_df)),
        "excluded_districts": skipped_districts
    },
    "notes": [
        "R2 is not classification accuracy.",
        "Negative R2 indicates poor fit relative to a mean baseline.",
        "Missing historical values were not filled with zeros.",
        "Forecasts are model estimates, not actual observations."
    ]
}

with open(
    BASE_DIR / "visualization_metrics.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        visualization_metrics,
        f,
        indent=4
    )

# ================================================================
# 18. VERIFY ALL GENERATED OUTPUTS
# ================================================================

output_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "comparison_results.csv",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "district_performance.csv",
    "visualization_metrics.json"
]

print("\n" + "=" * 70)
print("OUTPUT FILE VERIFICATION")
print("=" * 70)

for filename in output_files:

    path = BASE_DIR / filename

    if path.exists():
        print(f"[SAVED] {filename}")
    else:
        print(f"[NOT GENERATED] {filename}")

print("\nAll files are located in:")
print(BASE_DIR)

print("\nVISUALIZATION AND RESULTS GENERATION COMPLETED")
```
