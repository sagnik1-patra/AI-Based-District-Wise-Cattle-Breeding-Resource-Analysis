```python
from pathlib import Path
import os
import json
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml

from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")
np.random.seed(42)
tf.random.set_seed(42)

# ============================================================
# 1. PATH CONFIGURATION
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based District-Wise Cattle Breeding Resource Analysis"
)

DATASET_PATH = BASE_DIR / "Table_10.13_cowbulls_kept_AIcentres_0.csv"

BASE_DIR.mkdir(parents=True, exist_ok=True)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

print("=" * 65)
print("AI-BASED CATTLE BREEDING RESOURCE ANALYSIS")
print("=" * 65)

# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)

print("\nOriginal Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

# Remove entirely empty rows and columns.
df = df.dropna(how="all").dropna(axis=1, how="all")

district_col = "District/Year"

if district_col not in df.columns:
    raise ValueError(
        "Expected column 'District/Year' was not found."
    )

df[district_col] = df[district_col].astype("string").str.strip()

# Remove rows without a district name.
df = df.dropna(subset=[district_col])
df = df[df[district_col] != ""]

# Separate the Punjab aggregate to prevent double counting.
punjab_total = df[
    df[district_col].str.lower() == "punjab"
].copy()

df = df[
    df[district_col].str.lower() != "punjab"
].copy()

year_cols = sorted(
    [c for c in df.columns if str(c).isdigit()],
    key=lambda x: int(x)
)

if not year_cols:
    raise ValueError("No year columns found.")

for col in year_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("\nDistricts:", df[district_col].nunique())
print("Years:", year_cols[0], "to", year_cols[-1])

# ============================================================
# 3. TRANSFORM WIDE DATA INTO LONG FORMAT
# ============================================================

long_df = df.melt(
    id_vars=[district_col],
    value_vars=year_cols,
    var_name="Year",
    value_name="Animal_Count"
)

long_df = long_df.rename(
    columns={district_col: "District"}
)

long_df["Year"] = long_df["Year"].astype(int)
long_df["Animal_Count"] = pd.to_numeric(
    long_df["Animal_Count"],
    errors="coerce"
)

# Missing observations are excluded, not replaced with zero.
long_df = long_df.dropna(subset=["Animal_Count"])

long_df = long_df.sort_values(
    ["District", "Year"]
).reset_index(drop=True)

long_df.to_csv(
    BASE_DIR / "historical_data.csv",
    index=False
)

print("\nValid historical observations:", len(long_df))
print(long_df.head(10))

# ============================================================
# 4. FEATURE ENGINEERING
# ============================================================

# Use only previously observed consecutive years.
# This prevents lag features from silently crossing gaps.

LAGS = 3
FEATURES = [
    "Year",
    "Lag_1",
    "Lag_2",
    "Lag_3",
    "Moving_Average_3",
    "Previous_Change",
    "District_Code"
]

district_names = sorted(long_df["District"].unique())
district_mapping = {
    district: i
    for i, district in enumerate(district_names)
}

long_df["District_Code"] = (
    long_df["District"].map(district_mapping).astype(int)
)

records = []

for district, group in long_df.groupby("District"):
    group = group.sort_values("Year")
    counts_by_year = dict(
        zip(group["Year"], group["Animal_Count"])
    )

    for year in sorted(counts_by_year):
        previous_years = [
            year - 1,
            year - 2,
            year - 3
        ]

        if not all(y in counts_by_year for y in previous_years):
            continue

        lags = [
            float(counts_by_year[y])
            for y in previous_years
        ]

        records.append({
            "District": district,
            "Year": year,
            "Lag_1": lags[0],
            "Lag_2": lags[1],
            "Lag_3": lags[2],
            "Moving_Average_3": np.mean(lags),
            "Previous_Change": lags[0] - lags[1],
            "District_Code": district_mapping[district],
            "Animal_Count": float(counts_by_year[year])
        })

model_df = pd.DataFrame(records)

if len(model_df) < 20:
    raise ValueError(
        "Insufficient consecutive-year observations for training."
    )

print("\nModel Dataset Shape:", model_df.shape)

# ============================================================
# 5. CHRONOLOGICAL TRAIN / VALIDATION / TEST SPLIT
# ============================================================

# No random splitting of historical time-series rows.
# Training: earlier years
# Validation: intermediate years
# Testing: latest years

years = sorted(model_df["Year"].unique())

if len(years) < 9:
    raise ValueError("Not enough distinct years for evaluation.")

train_end_idx = max(1, int(len(years) * 0.65) - 1)
val_end_idx = max(
    train_end_idx + 1,
    int(len(years) * 0.80) - 1
)

train_end = years[train_end_idx]
val_end = years[val_end_idx]

train_df = model_df[model_df["Year"] <= train_end].copy()

val_df = model_df[
    (model_df["Year"] > train_end) &
    (model_df["Year"] <= val_end)
].copy()

test_df = model_df[
    model_df["Year"] > val_end
].copy()

if min(len(train_df), len(val_df), len(test_df)) == 0:
    raise ValueError("One of the chronological splits is empty.")

X_train = train_df[FEATURES]
y_train = train_df["Animal_Count"]

X_val = val_df[FEATURES]
y_val = val_df["Animal_Count"]

X_test = test_df[FEATURES]
y_test = test_df["Animal_Count"]

print("\nTraining:", X_train.shape)
print("Validation:", X_val.shape)
print("Testing:", X_test.shape)

print("Train through:", train_end)
print("Validation through:", val_end)
print("Test years:", sorted(test_df["Year"].unique()))

# ============================================================
# 6. RANDOM FOREST MODEL
# ============================================================

rf_model = RandomForestRegressor(
    n_estimators=250,
    max_depth=12,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

rf_val_pred = np.maximum(
    rf_model.predict(X_val), 0
)

rf_test_pred = np.maximum(
    rf_model.predict(X_test), 0
)

# Save fitted baseline model.
joblib.dump(
    rf_model,
    BASE_DIR / "cattle_breeding_model.pkl"
)

print("\nRandom Forest model saved.")

# ============================================================
# 7. FEATURE SCALING FOR NEURAL NETWORK
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

# Save preprocessing metadata.
preprocessing = {
    "scaler": scaler,
    "district_mapping": district_mapping,
    "features": FEATURES,
    "lags": LAGS,
    "train_end_year": train_end,
    "validation_end_year": val_end
}

joblib.dump(
    preprocessing,
    BASE_DIR / "preprocessing.pkl"
)

# ============================================================
# 8. DEEP LEARNING MODEL
# ============================================================

nn_model = Sequential([
    Input(shape=(len(FEATURES),)),
    Dense(64, activation="relu"),
    Dense(32, activation="relu"),
    Dropout(0.10),
    Dense(16, activation="relu"),
    Dense(1, activation="linear")
])

nn_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=25,
    restore_best_weights=True
)

history = nn_model.fit(
    X_train_scaled,
    y_train.to_numpy(),
    validation_data=(
        X_val_scaled,
        y_val.to_numpy()
    ),
    epochs=200,
    batch_size=16,
    callbacks=[early_stopping],
    verbose=0
)

# Save neural network in requested H5 format.
nn_model.save(
    str(BASE_DIR / "cattle_breeding_model.h5")
)

nn_val_pred = np.maximum(
    nn_model.predict(
        X_val_scaled,
        verbose=0
    ).flatten(),
    0
)

nn_test_pred = np.maximum(
    nn_model.predict(
        X_test_scaled,
        verbose=0
    ).flatten(),
    0
)

pd.DataFrame(history.history).to_csv(
    BASE_DIR / "training_history.csv",
    index=False
)

print("Neural network model saved.")

# ============================================================
# 9. MODEL EVALUATION
# ============================================================

def evaluate_model(actual, predicted):
    actual = np.asarray(actual, dtype=float)
    predicted = np.asarray(predicted, dtype=float)

    result = {
        "MAE": float(mean_absolute_error(actual, predicted)),
        "RMSE": float(
            np.sqrt(mean_squared_error(actual, predicted))
        ),
        "R2": float(
            r2_score(actual, predicted)
        ) if len(actual) > 1 else None
    }

    return result

validation_metrics = {
    "Random_Forest": evaluate_model(
        y_val, rf_val_pred
    ),
    "Neural_Network": evaluate_model(
        y_val, nn_val_pred
    )
}

test_metrics = {
    "Random_Forest": evaluate_model(
        y_test, rf_test_pred
    ),
    "Neural_Network": evaluate_model(
        y_test, nn_test_pred
    )
}

print("\nVALIDATION METRICS")
print(json.dumps(validation_metrics, indent=4))

print("\nTEST METRICS")
print(json.dumps(test_metrics, indent=4))

# Select model using validation MAE only.
best_model_name = min(
    validation_metrics,
    key=lambda name: validation_metrics[name]["MAE"]
)

print("\nSelected Model:", best_model_name)

# ============================================================
# 10. SAVE METRICS.JSON
# ============================================================

metrics_data = {
    "project": "AI-Based District-Wise Cattle Breeding Resource Analysis",
    "target": "Animal_Count",
    "dataset_start_year": int(min(years)),
    "dataset_end_year": int(max(years)),
    "train_end_year": int(train_end),
    "validation_end_year": int(val_end),
    "training_records": len(train_df),
    "validation_records": len(val_df),
    "testing_records": len(test_df),
    "validation_metrics": validation_metrics,
    "test_metrics": test_metrics,
    "selected_model": best_model_name,
    "selection_criterion": "Lowest validation MAE",
    "notes": [
        "Missing observations are not assumed to be zero.",
        "Punjab aggregate excluded from district training.",
        "Chronological train-validation-test split.",
        "R2 is coefficient of determination, not classification accuracy.",
        "Forecasts are experimental estimates, not observed counts."
    ]
}

with open(
    BASE_DIR / "metrics.json",
    "w",
    encoding="utf-8"
) as file:
    json.dump(metrics_data, file, indent=4)

# ============================================================
# 11. SAVE CONFIG.YAML
# ============================================================

config = {
    "project": {
        "name": "AI-Based District-Wise Cattle Breeding Resource Analysis",
        "dataset": DATASET_PATH.name,
        "output_directory": str(BASE_DIR),
        "target": "Animal_Count"
    },
    "data": {
        "first_year": int(min(years)),
        "last_year": int(max(years)),
        "districts": district_names,
        "district_mapping": district_mapping,
        "features": FEATURES,
        "lags": LAGS
    },
    "models": {
        "random_forest": {
            "n_estimators": 250,
            "max_depth": 12,
            "min_samples_leaf": 2
        },
        "neural_network": {
            "hidden_layers": [64, 32, 16],
            "learning_rate": 0.001,
            "batch_size": 16,
            "maximum_epochs": 200
        }
    },
    "evaluation": {
        "split_type": "chronological",
        "train_end_year": int(train_end),
        "validation_end_year": int(val_end),
        "metrics": ["MAE", "RMSE", "R2"],
        "selected_model": best_model_name
    },
    "artifacts": {
        "keras_model": "cattle_breeding_model.h5",
        "random_forest": "cattle_breeding_model.pkl",
        "preprocessing": "preprocessing.pkl",
        "metrics": "metrics.json"
    }
}

with open(
    BASE_DIR / "config.yaml",
    "w",
    encoding="utf-8"
) as file:
    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

# ============================================================
# 12. SAVE TEST RESULTS
# ============================================================

results = test_df[
    ["District", "Year", "Animal_Count"]
].copy()

results = results.rename(
    columns={"Animal_Count": "Actual_Count"}
)

results["RF_Prediction"] = rf_test_pred
results["NN_Prediction"] = nn_test_pred

results["Selected_Prediction"] = (
    rf_test_pred
    if best_model_name == "Random_Forest"
    else nn_test_pred
)

results["Absolute_Error"] = np.abs(
    results["Actual_Count"] -
    results["Selected_Prediction"]
)

results.to_csv(
    BASE_DIR / "model_results.csv",
    index=False
)

# ============================================================
# 13. FUTURE FORECASTING (2019-2023)
# ============================================================

# Recursive predictions are generated only for districts
# with observations in the final three dataset years.
# Districts lacking those values are excluded.
#
# These are retrospective forecasts, NOT observed data.
# No future-year actual values are used.

FORECAST_HORIZON = 5
last_dataset_year = max(int(c) for c in year_cols)

forecast_records = []
skipped_districts = []

def predict_selected(feature_dict):
    frame = pd.DataFrame([feature_dict])[FEATURES]

    if best_model_name == "Random_Forest":
        prediction = rf_model.predict(frame)[0]
    else:
        scaled = scaler.transform(frame)
        prediction = nn_model.predict(
            scaled,
            verbose=0
        )[0][0]

    return max(0.0, float(prediction))

for district, group in long_df.groupby("District"):

    history_values = dict(
        zip(group["Year"], group["Animal_Count"])
    )

    required_years = [
        last_dataset_year - 2,
        last_dataset_year - 1,
        last_dataset_year
    ]

    if not all(y in history_values for y in required_years):
        skipped_districts.append(district)
        continue

    history_values = {
        int(y): float(v)
        for y, v in history_values.items()
    }

    for future_year in range(
        last_dataset_year + 1,
        last_dataset_year + FORECAST_HORIZON + 1
    ):

        lag1 = history_values[future_year - 1]
        lag2 = history_values[future_year - 2]
        lag3 = history_values[future_year - 3]

        feature_dict = {
            "Year": future_year,
            "Lag_1": lag1,
            "Lag_2": lag2,
            "Lag_3": lag3,
            "Moving_Average_3": np.mean(
                [lag1, lag2, lag3]
            ),
            "Previous_Change": lag1 - lag2,
            "District_Code": district_mapping[district]
        }

        prediction = predict_selected(feature_dict)

        # Recursive forecasting:
        # Predicted values become subsequent lag values.
        history_values[future_year] = prediction

        forecast_records.append({
            "District": district,
            "Year": future_year,
            "Predicted_Animal_Count": prediction,
            "Rounded_Prediction": int(round(prediction)),
            "Model": best_model_name,
            "Prediction_Type": "Recursive forecast"
        })

forecast_df = pd.DataFrame(
    forecast_records,
    columns=[
        "District",
        "Year",
        "Predicted_Animal_Count",
        "Rounded_Prediction",
        "Model",
        "Prediction_Type"
    ]
)

forecast_df.to_csv(
    BASE_DIR / "future_predictions.csv",
    index=False
)

print("\nFuture Predictions:")
print(forecast_df.head(15))

if skipped_districts:
    print("\nDistricts excluded from future forecasting")
    print("(insufficient final-year observations):")
    print(skipped_districts)

# ============================================================
# 14. DISPLAY TRAINING HISTORY
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("Neural Network Training History")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "training_loss.png",
    dpi=300
)

plt.show()

# ============================================================
# 15. MODEL COMPARISON
# ============================================================

comparison = pd.DataFrame(test_metrics).T

print("\nModel Comparison:")
print(comparison)

comparison.to_csv(
    BASE_DIR / "model_comparison.csv"
)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

comparison["MAE"].plot(
    kind="bar",
    ax=axes[0],
    title="Test MAE Comparison"
)

comparison["RMSE"].plot(
    kind="bar",
    ax=axes[1],
    title="Test RMSE Comparison"
)

for ax in axes:
    ax.set_xlabel("Model")
    ax.tick_params(axis="x", rotation=0)
    ax.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "model_comparison.png",
    dpi=300
)

plt.show()

# ============================================================
# 16. ACTUAL VS PREDICTED
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    results["Actual_Count"],
    results["Selected_Prediction"],
    alpha=0.65
)

max_value = max(
    results["Actual_Count"].max(),
    results["Selected_Prediction"].max(),
    1
)

plt.plot(
    [0, max_value],
    [0, max_value],
    linestyle="--",
    label="Perfect Prediction"
)

plt.xlabel("Actual Animal Count")
plt.ylabel("Predicted Animal Count")
plt.title("Actual vs Predicted Cattle Breeding Resources")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "actual_vs_predicted.png",
    dpi=300
)

plt.show()

# ============================================================
# 17. FINAL FILE VERIFICATION
# ============================================================

required_files = [
    "cattle_breeding_model.h5",
    "cattle_breeding_model.pkl",
    "preprocessing.pkl",
    "config.yaml",
    "metrics.json",
    "historical_data.csv",
    "model_results.csv",
    "future_predictions.csv",
    "training_history.csv",
    "model_comparison.csv",
    "training_loss.png",
    "model_comparison.png",
    "actual_vs_predicted.png"
]

print("\n" + "=" * 65)
print("PROJECT OUTPUT VERIFICATION")
print("=" * 65)

for filename in required_files:
    path = BASE_DIR / filename
    status = "SAVED" if path.exists() else "MISSING"
    print(f"[{status}] {filename}")

print("\nSelected Model:", best_model_name)
print("\nAll outputs stored in:")
print(BASE_DIR)

print("\nPROJECT EXECUTION COMPLETED")
```
