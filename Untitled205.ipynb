```python
from pathlib import Path
import json
import time
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

SEED = 42
rng = np.random.default_rng(SEED)

# ================================================================
# 1. PROJECT CONFIGURATION
# ================================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based District-Wise Cattle Breeding Resource Analysis"
)

DATASET_PATH = (
    BASE_DIR / "Table_10.13_cowbulls_kept_AIcentres_0.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

print("=" * 75)
print("PSO-BASED CATTLE BREEDING RESOURCE ANALYSIS")
print("=" * 75)

# ================================================================
# 2. LOAD AND CLEAN DATA
# ================================================================

df = pd.read_csv(DATASET_PATH)

print("\nOriginal dataset shape:", df.shape)

df = df.dropna(how="all")
df = df.dropna(axis=1, how="all")

DISTRICT_COL = "District/Year"

if DISTRICT_COL not in df.columns:
    raise ValueError(
        "Required column District/Year not found."
    )

df[DISTRICT_COL] = (
    df[DISTRICT_COL]
    .astype("string")
    .str.strip()
)

df = df.dropna(subset=[DISTRICT_COL])
df = df[df[DISTRICT_COL] != ""].copy()

# Exclude statewide aggregate from district training.
df = df[
    df[DISTRICT_COL].str.lower() != "punjab"
].copy()

year_cols = sorted(
    [c for c in df.columns if str(c).isdigit()],
    key=int
)

if not year_cols:
    raise ValueError("No year columns found.")

for col in year_cols:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# ================================================================
# 3. WIDE TO LONG FORMAT
# ================================================================

long_df = df.melt(
    id_vars=[DISTRICT_COL],
    value_vars=year_cols,
    var_name="Year",
    value_name="Animal_Count"
)

long_df = long_df.rename(
    columns={DISTRICT_COL: "District"}
)

long_df["Year"] = long_df["Year"].astype(int)

long_df["Animal_Count"] = pd.to_numeric(
    long_df["Animal_Count"],
    errors="coerce"
)

# Keep genuine zero counts.
# Remove missing or invalid observations only.
long_df = long_df.dropna(
    subset=["Animal_Count"]
)

long_df = long_df[
    long_df["Animal_Count"] >= 0
].copy()

long_df = long_df.sort_values(
    ["District", "Year"]
).reset_index(drop=True)

print("\nDistricts:", long_df["District"].nunique())
print("Valid observations:", len(long_df))
print("Year range:", min(year_cols), "-", max(year_cols))

# ================================================================
# 4. FEATURE ENGINEERING
# ================================================================

district_names = sorted(
    long_df["District"].unique()
)

district_mapping = {
    district: i
    for i, district in enumerate(district_names)
}

FEATURES = [
    "Year",
    "Lag_1",
    "Lag_2",
    "Lag_3",
    "Moving_Average_3",
    "Previous_Change",
    "District_Code"
]

records = []

for district, group in long_df.groupby("District"):

    history = dict(
        zip(group["Year"], group["Animal_Count"])
    )

    for year in sorted(history):

        required_years = [
            year - 1,
            year - 2,
            year - 3
        ]

        # Do not create lags across missing years.
        if not all(
            y in history for y in required_years
        ):
            continue

        lag1 = float(history[year - 1])
        lag2 = float(history[year - 2])
        lag3 = float(history[year - 3])

        records.append({
            "District": district,
            "Year": int(year),
            "Lag_1": lag1,
            "Lag_2": lag2,
            "Lag_3": lag3,
            "Moving_Average_3": np.mean(
                [lag1, lag2, lag3]
            ),
            "Previous_Change": lag1 - lag2,
            "District_Code": district_mapping[district],
            "Animal_Count": float(history[year])
        })

model_df = pd.DataFrame(records)

if model_df.empty:
    raise ValueError(
        "No valid consecutive-year observations."
    )

print("\nModel dataset shape:", model_df.shape)

# ================================================================
# 5. CHRONOLOGICAL TRAIN / VALIDATION / TEST SPLIT
# ================================================================

# Same splitting rule as the previous AIS implementation.
# PSO only sees training and validation data.
# The test period is reserved for final evaluation.

years = sorted(model_df["Year"].unique())

if len(years) < 9:
    raise ValueError(
        "Not enough years for chronological splitting."
    )

train_end_idx = max(
    1,
    int(len(years) * 0.65) - 1
)

val_end_idx = max(
    train_end_idx + 1,
    int(len(years) * 0.80) - 1
)

train_end = years[train_end_idx]
val_end = years[val_end_idx]

train_df = model_df[
    model_df["Year"] <= train_end
].copy()

val_df = model_df[
    (model_df["Year"] > train_end) &
    (model_df["Year"] <= val_end)
].copy()

test_df = model_df[
    model_df["Year"] > val_end
].copy()

if min(
    len(train_df),
    len(val_df),
    len(test_df)
) == 0:
    raise ValueError("One of the data splits is empty.")

X_train = train_df[FEATURES]
y_train = train_df["Animal_Count"].to_numpy()

X_val = val_df[FEATURES]
y_val = val_df["Animal_Count"].to_numpy()

X_test = test_df[FEATURES]
y_test = test_df["Animal_Count"].to_numpy()

print("\nTraining records:", len(train_df))
print("Validation records:", len(val_df))
print("Testing records:", len(test_df))

print("\nTraining through:", train_end)
print("Validation through:", val_end)
print("Test years:", sorted(test_df["Year"].unique()))

# ================================================================
# 6. METRICS AND MODEL FUNCTIONS
# ================================================================

def evaluate(actual, predicted):

    actual = np.asarray(actual, dtype=float)
    predicted = np.asarray(predicted, dtype=float)

    mae = mean_absolute_error(
        actual,
        predicted
    )

    rmse = np.sqrt(
        mean_squared_error(
            actual,
            predicted
        )
    )

    r2 = (
        float(r2_score(actual, predicted))
        if len(actual) > 1 and np.var(actual) > 0
        else None
    )

    return {
        "MAE": float(mae),
        "RMSE": float(rmse),
        "R2": r2
    }


def create_rf(params):

    return RandomForestRegressor(
        n_estimators=int(
            params["n_estimators"]
        ),
        max_depth=int(
            params["max_depth"]
        ),
        min_samples_split=int(
            params["min_samples_split"]
        ),
        min_samples_leaf=int(
            params["min_samples_leaf"]
        ),
        max_features=float(
            params["max_features"]
        ),
        random_state=SEED,
        n_jobs=-1
    )

# ================================================================
# 7. BASELINE RANDOM FOREST
# ================================================================

baseline_params = {
    "n_estimators": 250,
    "max_depth": 12,
    "min_samples_split": 2,
    "min_samples_leaf": 2,
    "max_features": 1.0
}

baseline_model = create_rf(
    baseline_params
)

baseline_model.fit(
    X_train,
    y_train
)

baseline_val_pred = np.maximum(
    baseline_model.predict(X_val),
    0
)

baseline_val_mae = mean_absolute_error(
    y_val,
    baseline_val_pred
)

print("\nBaseline Validation MAE:", baseline_val_mae)

# ================================================================
# 8. PSO HYPERPARAMETER SEARCH SPACE
# ================================================================

PARAM_BOUNDS = {
    "n_estimators": (50, 300),
    "max_depth": (3, 20),
    "min_samples_split": (2, 15),
    "min_samples_leaf": (1, 10),
    "max_features": (0.4, 1.0)
}

PARAM_NAMES = list(
    PARAM_BOUNDS.keys()
)

INTEGER_PARAMS = {
    "n_estimators",
    "max_depth",
    "min_samples_split",
    "min_samples_leaf"
}

DIMENSIONS = len(PARAM_NAMES)

# Particle positions are normalized to [0, 1].
# This allows different parameter scales to be optimized
# with one velocity update rule.

def decode_particle(position):

    params = {}

    for i, name in enumerate(PARAM_NAMES):

        low, high = PARAM_BOUNDS[name]

        value = (
            low + float(position[i]) * (high - low)
        )

        if name in INTEGER_PARAMS:
            value = int(round(value))
        else:
            value = float(value)

        params[name] = value

    return params

# ================================================================
# 9. PSO FITNESS FUNCTION
# ================================================================

# Minimize validation MAE.
# Cache duplicate hyperparameter configurations.

fitness_cache = {}

def fitness(position):

    params = decode_particle(position)

    key = tuple(
        params[name]
        for name in PARAM_NAMES
    )

    if key in fitness_cache:
        return fitness_cache[key]

    model = create_rf(params)

    model.fit(
        X_train,
        y_train
    )

    prediction = np.maximum(
        model.predict(X_val),
        0
    )

    mae = float(
        mean_absolute_error(
            y_val,
            prediction
        )
    )

    fitness_cache[key] = mae

    return mae

# ================================================================
# 10. PARTICLE SWARM OPTIMIZATION
# ================================================================

N_PARTICLES = 12
N_ITERATIONS = 15

W_START = 0.9
W_END = 0.4

C1 = 1.8
C2 = 1.8

MAX_VELOCITY = 0.2

# Initialize particle positions and velocities.
positions = rng.uniform(
    0,
    1,
    size=(N_PARTICLES, DIMENSIONS)
)

velocities = rng.uniform(
    -0.1,
    0.1,
    size=(N_PARTICLES, DIMENSIONS)
)

# Personal best positions and scores.
pbest_positions = positions.copy()

pbest_scores = np.array([
    fitness(position)
    for position in positions
])

# Global best.
best_index = int(
    np.argmin(pbest_scores)
)

gbest_position = pbest_positions[
    best_index
].copy()

gbest_score = float(
    pbest_scores[best_index]
)

optimization_history = []

start_time = time.time()

print("\n" + "=" * 75)
print("STARTING PARTICLE SWARM OPTIMIZATION")
print("=" * 75)

for iteration in range(N_ITERATIONS):

    # Linearly decreasing inertia.
    w = W_START - (
        (W_START - W_END) *
        iteration / max(1, N_ITERATIONS - 1)
    )

    r1 = rng.random(
        size=(N_PARTICLES, DIMENSIONS)
    )

    r2 = rng.random(
        size=(N_PARTICLES, DIMENSIONS)
    )

    # PSO velocity update.
    cognitive = (
        C1 * r1 *
        (pbest_positions - positions)
    )

    social = (
        C2 * r2 *
        (gbest_position - positions)
    )

    velocities = (
        w * velocities +
        cognitive +
        social
    )

    # Velocity clamping.
    velocities = np.clip(
        velocities,
        -MAX_VELOCITY,
        MAX_VELOCITY
    )

    # Update positions.
    positions = positions + velocities

    # Keep particles within search space.
    positions = np.clip(
        positions,
        0,
        1
    )

    # Evaluate updated positions.
    scores = np.array([
        fitness(position)
        for position in positions
    ])

    # Update personal bests.
    improved = scores < pbest_scores

    pbest_scores[improved] = (
        scores[improved]
    )

    pbest_positions[improved] = (
        positions[improved].copy()
    )

    # Update global best.
    current_best_index = int(
        np.argmin(pbest_scores)
    )

    current_best_score = float(
        pbest_scores[current_best_index]
    )

    if current_best_score < gbest_score:

        gbest_score = current_best_score

        gbest_position = pbest_positions[
            current_best_index
        ].copy()

    optimization_history.append({
        "Iteration": iteration + 1,
        "Inertia": float(w),
        "Iteration_Best_MAE": float(
            np.min(scores)
        ),
        "Global_Best_MAE": float(
            gbest_score
        ),
        "Mean_Particle_MAE": float(
            np.mean(scores)
        ),
        "Evaluated_Configurations": len(
            fitness_cache
        )
    })

    print(
        f"Iteration {iteration + 1:02d}/{N_ITERATIONS}"
        f" | Best MAE: {gbest_score:.5f}"
        f" | Mean MAE: {np.mean(scores):.5f}"
        f" | Evaluations: {len(fitness_cache)}"
    )

elapsed_time = time.time() - start_time

best_params = decode_particle(
    gbest_position
)

print("\nPSO OPTIMIZATION COMPLETED")
print("Best parameters:", best_params)
print("Best validation MAE:", gbest_score)
print("Execution time:", round(elapsed_time, 2), "seconds")

history_df = pd.DataFrame(
    optimization_history
)

history_df.to_csv(
    BASE_DIR / "pso_optimization_history.csv",
    index=False
)

# ================================================================
# 11. PSO CONVERGENCE GRAPH
# ================================================================

plt.figure(figsize=(11, 6))

plt.plot(
    history_df["Iteration"],
    history_df["Global_Best_MAE"],
    marker="o",
    linewidth=2,
    label="PSO Best Validation MAE"
)

plt.plot(
    history_df["Iteration"],
    history_df["Mean_Particle_MAE"],
    linestyle=":",
    label="Mean Particle MAE"
)

plt.axhline(
    baseline_val_mae,
    linestyle="--",
    label="Baseline Validation MAE"
)

plt.xlabel("PSO Iteration")
plt.ylabel("Validation MAE")
plt.title(
    "PSO Optimization Convergence",
    fontsize=15,
    fontweight="bold"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "pso_convergence_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: pso_convergence_graph.png")

# ================================================================
# 12. TRAIN OPTIMIZED RANDOM FOREST
# ================================================================

# Fit on the original training data.
# This preserves the same evaluation protocol as AIS.

pso_model = create_rf(
    best_params
)

pso_model.fit(
    X_train,
    y_train
)

pso_val_pred = np.maximum(
    pso_model.predict(X_val),
    0
)

pso_test_pred = np.maximum(
    pso_model.predict(X_test),
    0
)

baseline_test_pred = np.maximum(
    baseline_model.predict(X_test),
    0
)

# Save PSO optimized model.
joblib.dump(
    pso_model,
    BASE_DIR / "pso_model.pkl"
)

preprocessing = {
    "features": FEATURES,
    "district_mapping": district_mapping,
    "train_end_year": int(train_end),
    "validation_end_year": int(val_end),
    "best_params": best_params,
    "model_type": "PSO Optimized Random Forest"
}

joblib.dump(
    preprocessing,
    BASE_DIR / "pso_preprocessing.pkl"
)

print("\nSaved: pso_model.pkl")
print("Saved: pso_preprocessing.pkl")

# ================================================================
# 13. MODEL PERFORMANCE EVALUATION
# ================================================================

baseline_metrics = evaluate(
    y_test,
    baseline_test_pred
)

pso_metrics = evaluate(
    y_test,
    pso_test_pred
)

comparison_df = pd.DataFrame([
    {
        "Model": "Baseline Random Forest",
        **baseline_metrics
    },
    {
        "Model": "PSO Optimized Random Forest",
        **pso_metrics
    }
])

comparison_df.to_csv(
    BASE_DIR / "pso_comparison_results.csv",
    index=False
)

print("\nMODEL COMPARISON")
print(comparison_df.to_string(index=False))

# ================================================================
# 14. PSO ACCURACY / PERFORMANCE GRAPH
# ================================================================

# R2 measures regression goodness-of-fit.
# It is not a classification accuracy percentage.
# Negative R2 scores are allowed.

fig, axes = plt.subplots(
    1, 2,
    figsize=(14, 5)
)

sns.barplot(
    data=comparison_df,
    x="Model",
    y="R2",
    ax=axes[0]
)

axes[0].set_title(
    "Test R² Score (Higher is Better)"
)

axes[0].set_ylabel("R² Score")
axes[0].set_xlabel("Model")
axes[0].tick_params(axis="x", rotation=12)
axes[0].axhline(0, color="black", linewidth=1)

sns.barplot(
    data=comparison_df,
    x="Model",
    y="MAE",
    ax=axes[1]
)

axes[1].set_title(
    "Test MAE (Lower is Better)"
)

axes[1].set_ylabel("Mean Absolute Error")
axes[1].set_xlabel("Model")
axes[1].tick_params(axis="x", rotation=12)

plt.suptitle(
    "PSO-Based Cattle Breeding Prediction Performance",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "pso_accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: pso_accuracy_graph.png")

# ================================================================
# 15. PSO CORRELATION HEATMAP
# ================================================================

heatmap_columns = [
    "Year",
    "Lag_1",
    "Lag_2",
    "Lag_3",
    "Moving_Average_3",
    "Previous_Change",
    "Animal_Count"
]

correlation = model_df[
    heatmap_columns
].corr()

plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    square=True
)

plt.title(
    "PSO Project - Feature Correlation Heatmap",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "pso_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: pso_heatmap.png")

# ================================================================
# 16. PSO MODEL COMPARISON GRAPH
# ================================================================

fig, axes = plt.subplots(
    1, 3,
    figsize=(17, 5)
)

metric_info = [
    ("MAE", "Mean Absolute Error"),
    ("RMSE", "Root Mean Squared Error"),
    ("R2", "R² Score")
]

for ax, (metric, title) in zip(
    axes,
    metric_info
):

    sns.barplot(
        data=comparison_df,
        x="Model",
        y=metric,
        ax=ax
    )

    ax.set_title(title)
    ax.set_xlabel("Model")
    ax.set_ylabel(metric)
    ax.tick_params(
        axis="x",
        rotation=15
    )

    if metric == "R2":
        ax.axhline(
            0,
            color="black",
            linewidth=1
        )

plt.suptitle(
    "Baseline Random Forest vs PSO Optimization",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "pso_comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: pso_comparison_graph.png")
print("Saved: pso_comparison_results.csv")

# ================================================================
# 17. SAVE PSO RESULT CSV
# ================================================================

result_df = test_df[
    ["District", "Year", "Animal_Count"]
].copy()

result_df = result_df.rename(
    columns={
        "Animal_Count": "Actual_Count"
    }
)

result_df["Baseline_Prediction"] = (
    baseline_test_pred
)

result_df["PSO_Prediction"] = (
    pso_test_pred
)

result_df["PSO_Absolute_Error"] = np.abs(
    result_df["Actual_Count"] -
    result_df["PSO_Prediction"]
)

result_df["PSO_Squared_Error"] = (
    result_df["Actual_Count"] -
    result_df["PSO_Prediction"]
) ** 2

result_df = result_df.sort_values(
    ["Year", "District"]
).reset_index(drop=True)

result_df.to_csv(
    BASE_DIR / "pso_result.csv",
    index=False
)

print("\nPSO RESULT CSV")
print(result_df.head(15).to_string(index=False))

print("\nSaved: pso_result.csv")

# ================================================================
# 18. PSO RESULT GRAPH
# ================================================================

fig, axes = plt.subplots(
    1, 2,
    figsize=(16, 6)
)

# Actual vs predicted scatter.
sns.scatterplot(
    data=result_df,
    x="Actual_Count",
    y="PSO_Prediction",
    ax=axes[0],
    alpha=0.7
)

max_value = max(
    result_df["Actual_Count"].max(),
    result_df["PSO_Prediction"].max(),
    1
)

axes[0].plot(
    [0, max_value],
    [0, max_value],
    "r--",
    label="Perfect Prediction"
)

axes[0].set_title(
    "PSO Actual vs Predicted"
)

axes[0].set_xlabel("Actual Animal Count")
axes[0].set_ylabel("Predicted Animal Count")
axes[0].legend()

# Year-wise total over available test observations.
year_summary = result_df.groupby(
    "Year",
    as_index=False
)[
    [
        "Actual_Count",
        "Baseline_Prediction",
        "PSO_Prediction"
    ]
].sum()

axes[1].plot(
    year_summary["Year"],
    year_summary["Actual_Count"],
    marker="o",
    label="Actual"
)

axes[1].plot(
    year_summary["Year"],
    year_summary["Baseline_Prediction"],
    marker="s",
    label="Baseline"
)

axes[1].plot(
    year_summary["Year"],
    year_summary["PSO_Prediction"],
    marker="^",
    label="PSO"
)

axes[1].set_title(
    "Year-wise Actual vs Predicted"
)

axes[1].set_xlabel("Year")
axes[1].set_ylabel("Animal Count")
axes[1].legend()

plt.suptitle(
    "PSO Cattle Breeding Resource Prediction Results",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "pso_result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: pso_result_graph.png")

# ================================================================
# 19. DISTRICT-WISE PSO PERFORMANCE
# ================================================================

district_results = []

for district, group in result_df.groupby("District"):

    actual = group["Actual_Count"].to_numpy()
    predicted = group["PSO_Prediction"].to_numpy()

    district_results.append({
        "District": district,
        "Observations": int(len(group)),
        **evaluate(actual, predicted)
    })

district_performance = pd.DataFrame(
    district_results
)

district_performance.to_csv(
    BASE_DIR / "pso_district_performance.csv",
    index=False
)

print("\nDISTRICT-WISE PSO PERFORMANCE")
print(district_performance.to_string(index=False))

# ================================================================
# 20. PSO FUTURE PREDICTIONS
# ================================================================

# Generate recursive forecasts for 2019-2023.
#
# Forecasts are generated only for districts with valid
# observed counts for 2016, 2017 and 2018.
#
# These forecasts are NOT actual observations.
#
# Recursive forecasting may accumulate errors.

LAST_YEAR = max(
    int(year)
    for year in year_cols
)

FORECAST_HORIZON = 5

prediction_records = []
skipped_districts = []

for district, group in long_df.groupby("District"):

    history = {
        int(row.Year): float(row.Animal_Count)
        for row in group.itertuples()
    }

    required_years = [
        LAST_YEAR - 2,
        LAST_YEAR - 1,
        LAST_YEAR
    ]

    if not all(
        year in history
        for year in required_years
    ):
        skipped_districts.append(district)
        continue

    for year in range(
        LAST_YEAR + 1,
        LAST_YEAR + FORECAST_HORIZON + 1
    ):

        lag1 = history[year - 1]
        lag2 = history[year - 2]
        lag3 = history[year - 3]

        feature_values = {
            "Year": year,
            "Lag_1": lag1,
            "Lag_2": lag2,
            "Lag_3": lag3,
            "Moving_Average_3": np.mean(
                [lag1, lag2, lag3]
            ),
            "Previous_Change": lag1 - lag2,
            "District_Code": district_mapping[district]
        }

        feature_df = pd.DataFrame(
            [feature_values]
        )[FEATURES]

        prediction = float(
            pso_model.predict(
                feature_df
            )[0]
        )

        prediction = max(
            0.0,
            prediction
        )

        # Store predicted value for next forecast step.
        history[year] = prediction

        prediction_records.append({
            "District": district,
            "Year": year,
            "PSO_Predicted_Count": prediction,
            "PSO_Rounded_Count": int(
                round(prediction)
            ),
            "Model": "PSO Optimized Random Forest",
            "Prediction_Type": "Recursive forecast"
        })

prediction_df = pd.DataFrame(
    prediction_records,
    columns=[
        "District",
        "Year",
        "PSO_Predicted_Count",
        "PSO_Rounded_Count",
        "Model",
        "Prediction_Type"
    ]
)

prediction_df.to_csv(
    BASE_DIR / "pso_prediction.csv",
    index=False
)

print("\nPSO FUTURE PREDICTIONS")
print(prediction_df.head(15).to_string(index=False))

print("\nSaved: pso_prediction.csv")

if skipped_districts:
    print(
        "\nDistricts excluded from forecasting "
        "because of missing 2016-2018 observations:"
    )
    print(skipped_districts)

# ================================================================
# 21. PSO PREDICTION GRAPH
# ================================================================

if not prediction_df.empty:

    fig, axes = plt.subplots(
        1, 2,
        figsize=(16, 6)
    )

    # Forecast totals for included districts.
    future_summary = prediction_df.groupby(
        "Year",
        as_index=False
    )["PSO_Predicted_Count"].sum()

    sns.lineplot(
        data=future_summary,
        x="Year",
        y="PSO_Predicted_Count",
        marker="o",
        ax=axes[0]
    )

    axes[0].set_title(
        "PSO Forecast by Year (Included Districts)"
    )

    axes[0].set_xlabel("Year")
    axes[0].set_ylabel("Predicted Animal Count")

    # Final-year district predictions.
    final_year = prediction_df["Year"].max()

    district_forecast = prediction_df[
        prediction_df["Year"] == final_year
    ].sort_values(
        "PSO_Predicted_Count",
        ascending=False
    )

    sns.barplot(
        data=district_forecast,
        x="PSO_Predicted_Count",
        y="District",
        ax=axes[1]
    )

    axes[1].set_title(
        f"District-wise PSO Forecast - {final_year}"
    )

    axes[1].set_xlabel(
        "Predicted Animal Count"
    )

    plt.suptitle(
        "PSO-Based Cattle Breeding Resource Forecast",
        fontsize=15,
        fontweight="bold"
    )

    plt.tight_layout()

    plt.savefig(
        BASE_DIR / "pso_prediction_graph.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    print("Saved: pso_prediction_graph.png")

else:
    print(
        "\nNo eligible districts for forecast graph."
    )

# ================================================================
# 22. SAVE PSO CONFIG.YAML
# ================================================================

config = {
    "project": {
        "name": "AI-Based District-Wise Cattle Breeding Resource Analysis",
        "algorithm": "Particle Swarm Optimization",
        "dataset": DATASET_PATH.name,
        "output_directory": str(BASE_DIR)
    },
    "dataset": {
        "first_year": int(min(years)),
        "last_year": int(max(years)),
        "target": "Animal_Count",
        "features": FEATURES,
        "district_mapping": district_mapping
    },
    "splitting": {
        "type": "chronological",
        "train_end_year": int(train_end),
        "validation_end_year": int(val_end),
        "training_records": int(len(train_df)),
        "validation_records": int(len(val_df)),
        "testing_records": int(len(test_df))
    },
    "pso": {
        "particles": N_PARTICLES,
        "iterations": N_ITERATIONS,
        "inertia_start": W_START,
        "inertia_end": W_END,
        "cognitive_coefficient": C1,
        "social_coefficient": C2,
        "maximum_velocity": MAX_VELOCITY,
        "fitness_function": "Validation MAE",
        "best_validation_mae": float(gbest_score),
        "best_parameters": best_params,
        "unique_configurations": int(
            len(fitness_cache)
        )
    },
    "forecast": {
        "start_year": int(LAST_YEAR + 1),
        "end_year": int(
            LAST_YEAR + FORECAST_HORIZON
        ),
        "excluded_districts": skipped_districts
    },
    "artifacts": {
        "model": "pso_model.pkl",
        "preprocessing": "pso_preprocessing.pkl",
        "metrics": "pso_metrics.json",
        "configuration": "pso_config.yaml"
    }
}

with open(
    BASE_DIR / "pso_config.yaml",
    "w",
    encoding="utf-8"
) as f:

    yaml.safe_dump(
        config,
        f,
        sort_keys=False,
        allow_unicode=True
    )

print("\nSaved: pso_config.yaml")

# ================================================================
# 23. SAVE PSO METRICS.JSON
# ================================================================

metrics = {
    "project": "PSO-Based Cattle Breeding Resource Analysis",
    "algorithm": "Particle Swarm Optimization",
    "optimization_objective": "Minimize validation MAE",
    "baseline_validation_mae": float(
        baseline_val_mae
    ),
    "pso_validation_mae": float(
        mean_absolute_error(
            y_val,
            pso_val_pred
        )
    ),
    "baseline_test_metrics": baseline_metrics,
    "pso_test_metrics": pso_metrics,
    "best_parameters": best_params,
    "iterations": N_ITERATIONS,
    "particles": N_PARTICLES,
    "unique_configurations_evaluated": int(
        len(fitness_cache)
    ),
    "optimization_seconds": float(
        elapsed_time
    ),
    "test_records": int(len(test_df)),
    "forecast_records": int(
        len(prediction_df)
    ),
    "notes": [
        "PSO optimizes Random Forest hyperparameters.",
        "Test data is not used during optimization.",
        "R2 is not classification accuracy.",
        "Missing historical values are not replaced with zero.",
        "Forecasts are experimental recursive projections."
    ]
}

with open(
    BASE_DIR / "pso_metrics.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metrics,
        f,
        indent=4
    )

print("Saved: pso_metrics.json")

# ================================================================
# 24. VERIFY GENERATED FILES
# ================================================================

output_files = [
    "pso_model.pkl",
    "pso_preprocessing.pkl",
    "pso_config.yaml",
    "pso_metrics.json",
    "pso_accuracy_graph.png",
    "pso_heatmap.png",
    "pso_comparison_graph.png",
    "pso_comparison_results.csv",
    "pso_result.csv",
    "pso_result_graph.png",
    "pso_prediction.csv",
    "pso_prediction_graph.png",
    "pso_convergence_graph.png",
    "pso_optimization_history.csv",
    "pso_district_performance.csv"
]

print("\n" + "=" * 75)
print("PSO OUTPUT FILE VERIFICATION")
print("=" * 75)

for filename in output_files:

    path = BASE_DIR / filename

    if path.exists():
        print(f"[SAVED] {filename}")
    else:
        print(f"[MISSING] {filename}")

# ================================================================
# 25. FINAL SUMMARY
# ================================================================

print("\n" + "=" * 75)
print("FINAL PSO RESULTS")
print("=" * 75)

print("\nBest PSO Hyperparameters:")
print(json.dumps(best_params, indent=4))

print("\nBaseline Test Metrics:")
print(json.dumps(baseline_metrics, indent=4))

print("\nPSO Test Metrics:")
print(json.dumps(pso_metrics, indent=4))

print("\nBest Validation MAE:", gbest_score)

print("\nOutput Directory:")
print(BASE_DIR)

print("\nPSO PROJECT EXECUTION COMPLETED SUCCESSFULLY")
```
